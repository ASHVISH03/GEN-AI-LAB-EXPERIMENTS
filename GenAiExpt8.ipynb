```python
from transformers import RobertaTokenizer, AutoModelForSeq2SeqLM

model_name = "Salesforce/codet5-base"

tokenizer = RobertaTokenizer.from_pretrained(model_name)
model = AutoModelForSeq2SeqLM.from_pretrained(model_name)

print("CodeT5 model loaded successfully.")
```

```python
prompt = "Write a Python function to calculate the factorial of a number."

inputs = tokenizer(
    prompt,
    return_tensors="pt",
    max_length=128,
    truncation=True
)

outputs = model.generate(
    **inputs,
    max_length=128,
    num_beams=5,
    early_stopping=True
)

generated_code = tokenizer.decode(
    outputs[0],
    skip_special_tokens=True
)

print("Generated Code:")
print(generated_code)
```

```python
prompt = "Write a Python program to check whether a number is prime."

inputs = tokenizer(
    prompt,
    return_tensors="pt",
    max_length=128,
    truncation=True
)

outputs = model.generate(
    **inputs,
    max_length=128,
    num_beams=5,
    early_stopping=True
)

generated_code = tokenizer.decode(
    outputs[0],
    skip_special_tokens=True
)

print("Natural Language Instruction:")
print(prompt)

print("\nGenerated Python Code:")
print(generated_code)
```

```python
faulty_code = """
def add_numbers(a, b):
    return a + c

print(add_numbers(5, 3))
"""

prompt = "Fix the following Python code:\n" + faulty_code

inputs = tokenizer(
    prompt,
    return_tensors="pt",
    max_length=256,
    truncation=True
)

outputs = model.generate(
    **inputs,
    max_length=256,
    num_beams=5,
    early_stopping=True
)

fixed_code = tokenizer.decode(
    outputs[0],
    skip_special_tokens=True
)

print("Faulty Code:")
print(faulty_code)

print("Fixed Code:")
print(fixed_code)
```

```python
faulty_code = """
def divide(a, b):
    result = a / 0
    return result

print(divide(10, 2))
"""

prompt = """Fix the Python code below. Return only the corrected Python code.

Code:
""" + faulty_code

inputs = tokenizer(
    prompt,
    return_tensors="pt",
    max_length=256,
    truncation=True
)

outputs = model.generate(
    **inputs,
    max_length=256,
    num_beams=10,
    no_repeat_ngram_size=2
)

fixed_code = tokenizer.decode(
    outputs[0],
    skip_special_tokens=True
)

print("Original Code:")
print(faulty_code)

print("\nAI Suggested Fix:")
print(fixed_code)
```

```python
faulty_code = """
def divide(a, b):
    result = a / 0
    return result

print(divide(10, 2))
"""

prompt = "fix: " + faulty_code

inputs = tokenizer(
    prompt,
    return_tensors="pt",
    max_length=256,
    truncation=True
)

outputs = model.generate(
    **inputs,
    max_length=256,
    num_beams=5
)

fixed_code = tokenizer.decode(
    outputs[0],
    skip_special_tokens=True
)

print("Faulty Code:")
print(faulty_code)

print("\nAI Fixed Code:")
print(fixed_code)
```
