```python
from transformers import pipeline

text_generator = pipeline(
    "text-generation",
    model="google/flan-t5-small"
)

print("Text generator loaded.")
```

```python
topic = "Generative Artificial Intelligence"

text = text_generator(
    f"Write a short paragraph about {topic}.",
    max_new_tokens=80
)[0]["generated_text"]

image = pipe(
    f"A futuristic illustration representing {topic}",
    num_inference_steps=5,
    height=256,
    width=256
).images[0]

image.save("generated_content_image.png")

print("GENERATED TEXT")
print("-" * 40)
print(text)
print("\nImage saved as generated_content_image.png")

display(image)
```
