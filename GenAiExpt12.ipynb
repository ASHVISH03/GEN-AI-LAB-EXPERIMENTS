```python
import os, re, time, getpass
import gradio as gr
from groq import Groq

api_key = os.getenv("GROQ_API_KEY") or getpass.getpass("Enter Groq API key: ")
client = Groq(api_key=api_key)

print("Groq client initialized successfully.")
```

```python
def generate(prompt):
    response = client.chat.completions.create(
        model="openai/gpt-oss-20b",
        messages=[{"role": "user", "content": prompt}],
        max_tokens=100
    ).choices[0].message.content
    return response

print(generate("Explain Generative AI in 3 sentences."))
```

```python
models = client.models.list()

for m in models.data:
    print(m.id)
```

```python
demo = gr.Interface(
    fn=generate,
    inputs="text",
    outputs="text",
    title="Generative AI Assistant"
)

demo.launch()
```
